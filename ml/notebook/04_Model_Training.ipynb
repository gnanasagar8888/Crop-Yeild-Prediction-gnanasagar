```python
import pandas as pd
import joblib
import os

from xgboost import XGBRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
)
```

```python
X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")

y_train = pd.read_csv("../data/processed/y_train.csv").squeeze()
y_test = pd.read_csv("../data/processed/y_test.csv").squeeze()

print(X_train.shape)
print(X_test.shape)
```

```python
model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=8,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
)
```

```python
model.fit(X_train, y_train)
```

```python
pred = model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, pred)
mse = mean_squared_error(y_test, pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, pred)

print("MAE :", mae)
print("RMSE:", rmse)
print("R2  :", r2)
```

```python
os.makedirs("../models", exist_ok=True)

joblib.dump(model, "../models/yield_model.pkl")

print("Model saved successfully.")
```

```python
from sklearn.preprocessing import LabelEncoder
import joblib
import os

# Load the ORIGINAL dataset (before encoding)
df = pd.read_csv("../data/final_dataset.csv")   # <-- change filename if needed

categorical_columns = ["crop", "season", "state"]

label_encoders = {}

for col in categorical_columns:
    le = LabelEncoder()
    le.fit(df[col].astype(str))      # IMPORTANT: fit on original strings
    label_encoders[col] = le

os.makedirs("../models", exist_ok=True)

joblib.dump(label_encoders, "../models/label_encoders.pkl")

print(label_encoders["crop"].classes_)
```
