```python
crop_clean = crop.copy()
weather_clean = weather.copy()
soil_clean = soil.copy()
faostat_clean = faostat.copy()
```

```python
print("Number of States:", crop_clean["state"].nunique())
print(crop_clean["state"].unique())
```

```python
print("Number of Crops:", crop_clean["crop"].nunique())
print(crop_clean["crop"].unique())
```

```python
crop_clean["state"] = crop_clean["state"].str.strip().str.title()

crop_clean["crop"] = crop_clean["crop"].str.strip().str.title()

crop_clean["season"] = crop_clean["season"].str.strip().str.title()
```

```python
weather_clean["state"] = weather_clean["state"].str.strip().str.title()

soil_clean["state"] = soil_clean["state"].str.strip().str.title()

faostat_clean["Area"] = faostat_clean["Area"].str.strip().str.title()

faostat_clean["Item"] = faostat_clean["Item"].str.strip().str.title()
```

```python
crop_weather = pd.merge(
    crop_clean,
    weather_clean,
    on=["state", "year"],
    how="left"
)
```
